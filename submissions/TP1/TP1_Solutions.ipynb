```python
employees = pd.read_csv(DATA + "employees.csv")
employees.info()
employees.head()
```

```python
print(employees.isna().sum())
```

```python
plt.figure(figsize=(8, 5))
sns.heatmap(employees.isna(), cbar=False)
plt.title("Missing values in employees.csv (white/light = missing)")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

```python
emp2 = employees.copy()

new_row = pd.DataFrame([["-", "na"] + [np.nan] * (emp2.shape[1] - 2)],
                       columns=emp2.columns)
emp2 = pd.concat([emp2, new_row], ignore_index=True)
emp2.tail(3)
```

```python
print(emp2.isna().sum())
```

```python
markers = ["-", "na"]
for col in emp2.select_dtypes(exclude="number").columns:
    cleaned = emp2[col].astype(str).str.strip().str.lower()
    emp2.loc[cleaned.isin(markers), col] = np.nan

print(emp2.isna().sum())
```

```python
from sklearn.impute import SimpleImputer

first_col = emp2.columns[0]
num_cols = emp2.select_dtypes(include="number").columns
cat_cols = [c for c in emp2.select_dtypes(exclude="number").columns if c != first_col]

emp_imputed = emp2.copy()
emp_imputed[num_cols] = SimpleImputer(strategy="median").fit_transform(emp2[num_cols])
emp_imputed[cat_cols] = SimpleImputer(strategy="most_frequent").fit_transform(emp2[cat_cols])
```

```python
comparison = pd.DataFrame({"before": emp2.isna().sum(),
                           "after": emp_imputed.isna().sum()})
comparison
```

```python
from sklearn.preprocessing import (StandardScaler, MinMaxScaler, MaxAbsScaler,
                                   RobustScaler, normalize, Binarizer)

X = np.array([[1, -1,  2],
              [2,  0,  0],
              [0,  1, -1]])

results = {
    "StandardScaler": StandardScaler().fit_transform(X),
    "MinMaxScaler":   MinMaxScaler().fit_transform(X),
    "MaxAbsScaler":   MaxAbsScaler().fit_transform(X),
    "RobustScaler":   RobustScaler().fit_transform(X),
    "normalize (L2)": normalize(X),
}

for name, res in results.items():
    print(name)
    print(np.round(res, 3), "\n")
```

```python
binary = Binarizer(threshold=0).fit_transform(X)
print(binary)
```

```python
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, OneHotEncoder

data = pd.DataFrame({
    "Feature 1": ["male", "female", "female", "male"],
    "Feature 2": ["US", "Europe", "Asia", "Africa"],
    "Feature 3": ["Safari", "Safari", "Firefox", "Chrome"],
    "Target":    ["US", "France", "Japan", "Algeria"],
})

features = data[["Feature 1", "Feature 2", "Feature 3"]]
ord_enc = OrdinalEncoder()
features_encoded = ord_enc.fit_transform(features)
print(features_encoded)
print(ord_enc.categories_)
```

```python
label_enc = LabelEncoder()
y_encoded = label_enc.fit_transform(data["Target"])
print(y_encoded)
for i, name in enumerate(label_enc.classes_):
    print(name, "->", i)
```

```python
onehot = OneHotEncoder()
y_onehot = onehot.fit_transform(data[["Target"]]).toarray()
print(onehot.categories_)
print(y_onehot)
```
